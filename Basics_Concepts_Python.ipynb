### Arithmetic Operators

* Addition `x + y`
* Subtraction `x - y`
* Multiplication `x * y`
* Division `x / y`
* Floor Division `x // y`
* Exponentiation `x ** y`
* Square Root `x ** 0,5`
* Modulo `x % y`
* Convert to Integer `int(x)`
* Convert to Float `float(x)`

````python
#Arithmetic Operators Examples
x = 10
y = 64

print('Addition:', x + y)
print('Subtraction:', x - y)
print('Multiplication:', x * y)
print('Division:', x / y)
print('Floor Division:', x // y)
print('Modulo:', x % y)
print('Exponentiation:', x ** 2)
print('Square Root of Y:', y ** 0.5)
print('To Integer:', int(4.36))
print('To Float:', float(9))
````

### Rounding, Floor, Ceiling, and Absolute
* Rounding `round()`
* Ceiling - `ceil(x)` will return the smallest integer **greater than** or equal to `x`.
* Floor - `floor(x)` will return the largest integer **less than** or equal to `x`.
* Absolute - `abs(x)` will always return a **positive** number `x`

````python
import math
print('Round:', round(3.33))
print('Round with 3 Cases:', round(3.38156487, 3))
print('Ceiling:', math.ceil(3.001))
print('Floor:', math.floor(3.99))
print('Absolute:', abs(-3.15))
````

### Relational Operators

A condition is the establishment of a logical relationship between two elements. To define a logical relationship, it is necessary to use relational operators. A logical relationship (comparison) returns only the values true (`True`) or false (`False`).

* Equal to: `x == y` (there is no `x === y`)
* Greater than: `x > y`
* Less than: `x < y`
* Greater than or equal to: `x >= y`
* Less than or equal to: `x <= y`
* Not equal to: `x != y`

````python
#Relational Operators Examples
x = 3
y = 4

print('Equal to:', x == y)
print('Greater than:', x > y)
print('Less than:', x < y)
print('Greater than or Equal to:', x >= y)
print('Less than or Equal to:', x <= y)
print('Not Equal to:', x != y)
````

### Logical Operators
They are used for the same decision-making that requires more than one condition. They are also known as boolean operators. Logical operators only return true or false values.

| Logical Operator | Function    | Precedence |
|-----------------|-----------|---------------------|
| `NOT`           | Negation  | 1                   |
| `AND`           | Conjunction| 2                   |
| `OR`            | Disjunction| 3                   |

### Truth Tables for Logical Operators

Truth tables for the logical operators `AND`, `OR`, and `NOT`:

**AND (Conjunction)**

| A     | B     | A AND B |
|-------|-------|---------|
| True  | True  | True    |
| True  | False | False   |
| False | True  | False   |
| False | False | False   |

The `AND` operator returns `True` only if both operands are `True`. Otherwise, it returns `False`.

**OR (Disjunction)**

| A     | B     | A OR B |
|-------|-------|--------|
| True  | True  | True   |
| True  | False | True   |
| False | True  | True   |
| False | False | False  |

The `OR` operator returns `True` if at least one of the operands is `True`. It returns `False` only if both operands are `False`.

**NOT (Negation)**

| A     | NOT A |
|-------|-------|
| True  | False |
| False | True  |

The `NOT` operator negates the value of the operand. If the operand is `True`, it returns `False`, and if the operand is `False`, it returns `True`.

````python
#Logical Operators Examples
x = 1
y = 5

print((x > y) and (y > x))
print((x < y) or (y < x))
print(not (x < y))
````
