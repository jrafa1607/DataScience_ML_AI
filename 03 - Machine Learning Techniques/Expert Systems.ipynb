### **Expert Systems**

**Expert Systems** are computer programs designed to emulate the decision-making ability of a human expert. They are a branch of artificial intelligence that aims to solve complex problems by reasoning through bodies of knowledge, represented primarily as `if-then` rules rather than through conventional procedural code. Their main purpose is to assist humans in complex decision-making tasks, providing advice, explanations, and diagnoses in specific domains.

#### **Key Concepts**

*   **Knowledge Base:** Contains domain-specific knowledge, facts, and heuristic rules collected from human experts.

*   **Inference Engine:** The 'brain' of the system, which processes the knowledge base to draw conclusions and make decisions.

* **Working Memory:** Stores facts relevant to the current problem being solved.

* **User Interface:** Allows interaction between the user and the system.

* **Explanation Facility:** Justifies the system's reasoning process and conclusions to the user.

* **Knowledge Acquisition Module:** Used by domain experts to input or update knowledge in the knowledge base.

#### Technical Concepts: **Knowledge Base**

* The **Knowledge Base** is the repository of an expert system's domain-specific knowledge. It contains facts, rules, and other information related to the problem domain. Its purpose is to store all the knowledge that the system needs to reason and make decisions, much like a human expert's accumulated expertise.

  *   **Facts**: Specific pieces of information about the current situation or problem (e.g., "patient has a fever").
  *   **Rules**: Conditional statements that describe relationships and prescribe actions or conclusions based on certain conditions (e.g., "IF patient has fever AND patient has cough THEN conclude patient has flu").

* **Knowledge Representation** refers to the methods used to encode the knowledge in the knowledge base so that it can be effectively used by the inference engine. Common methods include:

  *   **Rules (If-Then Statements)**: The most common form, expressing knowledge in terms of conditions and consequents
  (e.g., IF `temperature > 100°F` THEN `diagnosis is fever`).

  *   **Frames**: A data structure that represents a stereotyped situation, object, or event (attributes) (e.g, A "Car" frame might have attributes like `model`, `color`, `engine_type`).
  
  *   **Semantic Networks**: A graphical representation where knowledge is represented as a network of interconnected nodes.

* The **Inference Engine** is the **"brain"** of the expert system. It processes the information in the knowledge base and working memory to draw conclusions or make decisions. It applies logical rules and reasoning strategies to infer new facts from existing ones. Two primary mechanisms are:
  * **Forward Chaining or Data-Driven**: Starts with known facts and applies rules to deduce new facts until a conclusion is reached or no more rules can be fired. It moves from data to conclusions
  * **Backward Chaining or Goal-Driven**: Starts with a desired goal or hypothesis and works backward to find the facts that would support that goal.

* **Working Memory ( or Short-term memory )**: Is a temporary storage area that holds the current state of the problem-solving process. It stores facts and hypotheses that are currently being considered or have been inferred during a specific session.

* **User Interface**: Provides the means for human users to interact with the expert system. It allows users to input information (e.g., symptoms, questions), receive system outputs (e.g., diagnoses, recommendations), and generally communicate with the system in an understandable way. It should be intuitive and user-friendly.

```python
#Expert Systems - Example of Knowledge Base ( Animals Classification )
knowledge_base = [
    {"conditions": ["has_hair", "gives_milk"], "conclusion": "is_mammal"},
    {"conditions": ["has_feathers", "lays_eggs"], "conclusion": "is_bird"},
    {"conditions": ["has_scales", "lays_eggs"], "conclusion": "is_reptile"},
    {"conditions": ["is_cold_blooded", "lives_in_water", "has_gills"], "conclusion": "is_fish"},
    {"conditions": ["is_cold_blooded", "lives_in_water", "lives_on_land"], "conclusion": "is_amphibian"},
    {"conditions": ["is_mammal", "eats_meat"], "conclusion": "is_carnivore_mammal"},
    {"conditions": ["is_mammal", "eats_plants"], "conclusion": "is_herbivore_mammal"},
    {"conditions": ["is_mammal", "has_hooves"], "conclusion": "is_ungulate"},
    {"conditions": ["is_bird", "flies"], "conclusion": "is_flying_bird"},
    {"conditions": ["is_bird", "swims"], "conclusion": "is_swimming_bird"},
    {"conditions": ["is_fish", "has_fins"], "conclusion": "is_aquatic_fish"},
    {"conditions": ["is_carnivore_mammal", "has_mane"], "conclusion": "identified_as_lion"},
    {"conditions": ["is_herbivore_mammal", "has_long_neck"], "conclusion": "identified_as_giraffe"},
    {"conditions": ["is_flying_bird", "sings"], "conclusion": "identified_as_sparrow"},
    {"conditions": ["is_swimming_bird", "waddles"], "conclusion": "identified_as_penguin"},
    {"conditions": ["is_reptile", "has_shell"], "conclusion": "identified_as_turtle"},
    {"conditions": ["is_amphibian", "jumps"], "conclusion": "identified_as_frog"}
]
```

```python
#Create the Inference Function ( Engine )

def inference_engine(facts, knowledge_base):
  # Start with the initial facts
    deduced_facts = set(facts)
    new_facts_inferred_in_cycle = True

    # Keep inferring until no new facts are deduced in a complete pass
    while new_facts_inferred_in_cycle:
        new_facts_inferred_in_cycle = False
        for rule in knowledge_base:
            conditions_met = True
            for condition in rule['conditions']:
                if condition not in deduced_facts:
                    conditions_met = False
                    break

            # If all conditions for the rule are met and the conclusion is not yet a fact
            if conditions_met and rule['conclusion'] not in deduced_facts:
                deduced_facts.add(rule['conclusion'])
                print(f"  Rule Fired: IF {' AND '.join(rule['conditions'])} THEN {rule['conclusion']}")
                new_facts_inferred_in_cycle = True # A new fact was inferred, so try another pass

    return sorted(list(deduced_facts))
```

```python
#Example 01
initial_facts = ["has_hair", "gives_milk", "eats_meat", "has_mane"]
deduced_facts = inference_engine(initial_facts, knowledge_base)
final_identification = [fact for fact in deduced_facts if fact.startswith('identified_as_')]

if final_identification:
    print(f"\nFinal Identification: {final_identification[0].replace('identified_as_', '')}")
else:
    print("\nNo specific animal identification could be made.")
```
