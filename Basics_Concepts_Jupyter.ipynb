## Cell Commands

* To run the cell press `Shift + Enter`
* Command Mode `Esc`
* Edit Mode `Enter`
* To insert a new cell press `a` (for **before** the selected cell) or `b` (for **after** the selected cell)
* To delete a cell press `d` twice (delete)
* Copy selected cell(s) press `c` (copy)
* Paste copied cell(s) press `v` (paste)
* Change type:
  * Code `y`
  * Markdown `m`

### Jupyter Magic

These are [Interactions](https://ipython.readthedocs.io/en/stable/interactive/magics.html) of the Jupyter IDE with the Cells. For example:

* `%timeit`: Measures the execution time of a `Python` **expression**
* `%%timeit`: Measures the execution time of a **cell**
